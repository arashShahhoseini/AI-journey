- پیاده‌سازی **Gradient Descent** بدون کتابخانه
- آماده‌سازی برای مدل‌های پیچیده‌تر

## 🧱 مرحله 1: مسئله را ساده شروع می‌کنیم

فرض کن داده‌های زیر را داریم:

| x | y |
|---|---|
| 1 | 2 |
| 2 | 4 |
| 3 | 6 |
| 4 | 8 |

✅ این داده‌ها از قانون ساده `y = 2x` ساخته شده‌اند.

ما می‌خواهیم مدل یاد بگیرد که `w` و `b` را طوری پیدا کند که:

```
y_pred = w * x + b
```

تا حد امکان به `y` واقعی نزدیک باشد.

</div>

````python
# داده‌ها
X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

# مقدار اولیه برای وزن و بایاس
w = 0.5
b = 0.0

# نرخ یادگیری (قدم برداشتن در مسیر کاهش خطا)
lr = 0.01

# تعداد دفعات یادگیری
epochs = 1000

# تابع پیش‌بینی: خروجی مدل را محاسبه می‌کند
def predict(X, w, b):
    return [w * x + b for x in X]

# محاسبه dw و db (شیب خطا نسبت به w و b)
def compute_gradients(X, Y, w, b):
    n = len(X)
    Y_pred = predict(X, w, b)
    
    dw = (2 / n) * sum((Y_pred[i] - Y[i]) * X[i] for i in range(n))
    db = (2 / n) * sum(Y_pred[i] - Y[i] for i in range(n))
    
    return dw, db

# حلقه آموزش
for epoch in range(epochs):
    dw, db = compute_gradients(X, Y, w, b)
    
    w = w - lr * dw
    b = b - lr * db
    
    if epoch % 200 == 0:
        print(f"epoch {epoch:3d}: w={w:.4f}, b={b:.4f}, dw={dw:.4f}, db={db:.4f}")
````

<div dir='rtl'>

### 📘 توضیح ریاضی ساده

- `dw` یعنی: اگر وزن را کمی تغییر دهیم، خطا چقدر تغییر می‌کند؟
- `db` یعنی: اگر بایاس را کمی تغییر دهیم، خطا چقدر تغییر می‌کند؟
- Gradient Descent یعنی: برعکس جهت شیب حرکت می‌کنیم تا خطا کم شود.

</div>

<div dir='rtl'>

## 🧪 مرحله 2: تست مدل بعد از یادگیری

</div>

````python
test_value = 123
prediction = w * test_value + b
print(f"پیش‌بینی مدل برای x=123: {prediction:.2f}")
print(f"مقدار واقعی (ideal): {2 * test_value}")
print(f"\n✅ وزن نهایی (w): {w:.4f} (مقدار ایده‌آل: 2)")
print(f"✅ بایاس نهایی (b): {b:.4f} (مقدار ایده‌آل: 0)")
````

## 🚀 مرحله 3: گسترش به داده‌های چندبعدی (پیشرفته‌تر)

- حالا فرض کن به جای `x`، چند ویژگی داریم: مثلاً مساحت و تعداد اتاق‌ها → قیمت خانه.

</div>

````python
# داده‌های ساده با 2 ویژگی
X_multi = [[1, 2],
           [2, 3],
           [3, 4],
           [4, 5]]

Y_multi = [5, 7, 9, 11]  # رابطه y = x1 + 2*x2

# وزن‌ها و بایاس
w_multi = [0.0, 0.0]
b_multi = 0.0
lr_multi = 0.01
epochs_multi = 1000

def predict_multi(X, w, b):
    return [sum(w[j] * sample[j] for j in range(len(w))) + b for sample in X]

def compute_gradients_multi(X, Y, w, b):
    n = len(X)
    Y_pred = predict_multi(X, w, b)
    dw = [0.0] * len(w)
    db = 0.0
    
    for j in range(len(w)):
        dw[j] = (2 / n) * sum((Y_pred[i] - Y[i]) * X[i][j] for i in range(n))
    
    db = (2 / n) * sum(Y_pred[i] - Y[i] for i in range(n))
    
    return dw, db

print("آموزش مدل چندبعدی:")
print("-" * 40)

for epoch in range(epochs_multi):
    dw, db = compute_gradients_multi(X_multi, Y_multi, w_multi, b_multi)
    
    for j in range(len(w_multi)):
        w_multi[j] -= lr_multi * dw[j]
    
    b_multi -= lr_multi * db
    
    if epoch % 200 == 0:
        print(f"epoch {epoch:3d}: w={[round(wi, 4) for wi in w_multi]}, b={b_multi:.4f}")
````

````python
# تست مدل چندبعدی
sample = [5, 6]
pred_multi = sum(w_multi[j] * sample[j] for j in range(len(w_multi))) + b_multi
print(f"پیش‌بینی برای نمونه {sample}: {pred_multi:.2f}")
print(f"مقدار واقعی (ideal): {sample[0] + 2*sample[1]}")
print(f"\n✅ وزن‌های نهایی: w1={w_multi[0]:.4f}, w2={w_multi[1]:.4f}")
print(f"✅ بایاس نهایی: {b_multi:.4f}")
````
